```python
import numpy as np
import os


# Datos de prueba
data3 = np.array([[1, 2, 3], [4, 5, 6]])

# Guardar con encabezado y formato limpio
np.savetxt(
    "out.csv",
    data3,
    delimiter=",",
    fmt="%d",
    header="Columna1,Columna2,Columna3",
    comments="",
)

print(os.getcwd())
```
