```python
# PROBLEMA 1: Dos Sumas
def dos_sumas(nums, target):
    vistos = {}
    for i, num in enumerate(nums):
        complemento = target - num
        if complemento in vistos:
            return [vistos[complemento], i]
        vistos[num] = i
    return []

# PROBLEMA 2: Validar Palíndromo
def es_palindromo(cadena):
    filtrado = "".join(c.lower() for c in cadena if c.isalnum())
    return filtrado == filtrado[::-1]

# PROBLEMA 3: Serie de Fibonacci
def fibonacci(n):
    if n <= 0: return []
    if n == 1: return [0]
    resultado = [0, 1]
    while len(resultado) < n:
        resultado.append(resultado[-1] + resultado[-2])
    return resultado

# === EJECUCIÓN Y PRUEBAS ===
print("--- EJECUTANDO PRUEBAS DE ALGORITMOS ---")
print(f"1. Dos Sumas para [2, 7, 11, 15] (Target: 9) -> Resultado: {dos_sumas([2, 7, 11, 15], 9)}")
print(f"2. ¿Es palíndromo 'Anita lava la tina'?        -> Resultado: {es_palindromo('Anita lava la tina')}")
print(f"3. Primeros 5 números de Fibonacci           -> Resultado: {fibonacci(5)}")
print("---------------------------------------")
```
