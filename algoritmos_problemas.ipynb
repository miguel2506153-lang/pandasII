### 1. Dos Sumas (Two Sum)
**Descripción:** Dado un arreglo de números enteros `nums` y un valor objetivo `target`, retorna los índices de los dos números que sumen el objetivo.
**Ejemplo:** `nums = [2, 7, 11, 15]`, `target = 9` -> Retorna `[0, 1]`

```python
def dos_sumas(nums, target):
    vistos = {}
    for i, num in enumerate(nums):
        complemento = target - num
        if complemento in vistos:
            return [vistos[complemento], i]
        vistos[num] = i
    return []

# Prueba tu solución
print(dos_sumas([2, 7, 11, 15], 9))
```

### 2. Validar Palíndromo
**Descripción:** Determina si una cadena de texto es un palíndromo, ignorando espacios, mayúsculas y caracteres no alfanuméricos.
**Ejemplo:** `"Anita lava la tina"` -> Retorna `True`

```python
def es_palindromo(cadena):
    filtrado = "".join(c.lower() for c in cadena if c.isalnum())
    return filtrado == filtrado[::-1]

# Prueba tu solución
print(es_palindromo("Anita lava la tina"))
```

### 3. Serie de Fibonacci
**Descripción:** Crea una función que devuelva los primeros `n` números de la serie de Fibonacci.
**Ejemplo:** `n = 5` -> Retorna `[0, 1, 1, 2, 3]`

```python
def fibonacci(n):
    if n <= 0: return []
    if n == 1: return [0]
    resultado = [0, 1]
    while len(resultado) < n:
        resultado.append(resultado[-1] + resultado[-2])
    return resultado

# Prueba tu solución
print(fibonacci(5))
```
